```python
france_pivot=pd.pivot_table(data=france_data,values="Quantity",index="InvoiceNo",columns="Description").fillna(0)
```

```python
france_encodded=france_pivot.applymap(lambda x:1 if x>0 else 0)
```

```python
pip install mlxtend
```

```python
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
frequent_itemset=apriori(france_encodded,min_support=0.05,use_colnames=True)
```

```python
frequent_itemset
```

```python
best_association = association_rules(frequent_itemset,)
```
